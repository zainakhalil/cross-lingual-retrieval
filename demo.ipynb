# Cross-lingual retrieval demo

An English query searches documents in Arabic, Spanish and English using one multilingual embedding model ([multilingual-E5 base](https://huggingface.co/intfloat/multilingual-e5-base)). The relevant document is in Arabic and shares no words with the query.

```python
!pip install -q sentence-transformers
```

```python
from sentence_transformers import SentenceTransformer, util

model = SentenceTransformer("intfloat/multilingual-e5-base")
```

```python
ar_doc = "اللقاحات تدرّب جهاز المناعة على التعرف على الفيروسات"  # Vaccines train the immune system to recognize viruses
es_doc = "La paella lleva arroz y azafrán"  # Paella has rice and saffron
en_doc = "The stock market fell on Monday"

# E5 models expect these prefixes
query = "query: how do vaccines work"
docs = ["passage: " + d for d in (ar_doc, es_doc, en_doc)]
```

```python
q = model.encode(query, normalize_embeddings=True)
D = model.encode(docs, normalize_embeddings=True)
print(util.cos_sim(q, D))
```

## Result

| Document | Language | Cosine similarity |
| --- | --- | --- |
| Vaccines train the immune system... | Arabic | **0.822** |
| Paella has rice and saffron | Spanish | 0.723 |
| The stock market fell on Monday | English | 0.715 |

The Arabic document ranks first with zero words in common with the query. A keyword ranker like BM25 would give it a score of 0.

Note: E5 scores bunch between about 0.7 and 0.9, so the ranking matters, not the raw value.
